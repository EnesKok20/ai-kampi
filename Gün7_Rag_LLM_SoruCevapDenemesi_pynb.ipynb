```python
bilgi_tabani=[
     "Python, 1991 yılında Guido van Rossum tarafından geliştirilen, okunabilirliği yüksek bir programlama dilidir.",
    "PyTorch, Facebook (Meta) tarafından geliştirilen, derin öğrenme projelerinde yaygın kullanılan bir kütüphanedir.",
    "CNN (Convolutional Neural Network), özellikle görüntü işleme problemlerinde kullanılan bir derin öğrenme mimarisidir.",
    "Transformer mimarisi, 2017 yılında 'Attention is All You Need' makalesiyle tanıtılmış ve LLM'lerin temelini oluşturmuştur.",
    "Pandas, Python'da veri analizi ve manipülasyonu için kullanılan popüler bir kütüphanedir.",
    "Titanic veri seti, makine öğrenmesi eğitiminde en çok kullanılan başlangıç veri setlerinden biridir."
]
for i, metin in enumerate(bilgi_tabani):
  print(f"{i}:{metin}")
```

enumerate(bilgi_tabani) fonkiyonu bizim metnimizi satır satır ayırmayı sağlıyor

```python
!pip install sentence-transformers --quiet
from sentence_transformers import SentenceTransformer

model_embed = SentenceTransformer('all-MiniLM-L6-v2')
embeddings = model_embed.encode(bilgi_tabani)

print(f"embedding boyutu: {embeddings.shape} ")
print(f"\nİlk cümlenin embedding vektörü (ilk 10 sayı):\n{embeddings[0][:10]}")
```

.encode(...) fonksiyonu, verdiğimiz 6 cümlelik listeyi (bilgi_tabani) tek tek alıp, her birini bir sayı vektörüne çeviriyor. Yani her cümle artık "Python, 1991 yılında..." gibi bir yazı değil, [0.23, -0.51, 0.88, ...] gibi bir sayı dizisi haline geliyor





.shape, bize elimizdeki verinin boyutunu gösterecek. Muhtemelen (6, 384) gibi bir sonuç çıkacak: 6 → bizim 6 cümlemiz, 384 → her cümlenin kaç sayıdan oluşan bir vektörle temsil edildiği (bu, all-MiniLM-L6-v2 modelinin sabit bir özelliği).

```python
soru = "PyTorch'u kim geliştirdi?"

soru_embedding = model_embed.encode([soru])

from sklearn.metrics.pairwise import cosine_similarity

benzerlikler = cosine_similarity(soru_embedding, embeddings)

print("Her cümleyle benzerlik skorları:")
for i, skor in enumerate(benzerlikler[0]):
    print(f"{i}: {skor:.4f} → {bilgi_tabani[i]}")
```

```python
en_yuksek_index = benzerlikler[0].argmax()
en_alakali_cumle = bilgi_tabani[en_yuksek_index]

print(f"Soru: {soru}")
print(f"En alakalı bulunan bilgi: {en_alakali_cumle}")
print(f"Benzerlik skoru: {benzerlikler[0][en_yuksek_index]:.4f}")
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-base")
llm_model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-base")

prompt = f"Context: {en_alakali_cumle}\nQuestion: {soru}\nAnswer:"

girisler = tokenizer(prompt, return_tensors="pt")
cikti = llm_model.generate(**girisler, max_length=50)

cevap = tokenizer.decode(cikti[0], skip_special_tokens=True)

print(f"Soru: {soru}")
print(f"Kullanılan bağlam (context): {en_alakali_cumle}")
print(f"\nLLM'in ürettiği cevap: {cevap}")
```

# RAG ile LLM Tabanlı Soru-Cevap Denemesi

Bu çalışmada, bir dil modeli (LLM) yüklenerek **RAG** (Retrieval-Augmented Generation: "getirmeyle-güçlendirilmiş üretim") mantığında bir soru-cevap sistemi kuruldu. Amaç, modele bir soru sorarken yanında ilgili bir **bağlam** (context) metni vererek cevabın bu bağlamdan üretilmesini sağlamak ve bağlamın cevaba etkisini gözlemlemekti.

## Amaç

- Bir LLM'i Hugging Face üzerinden yükleyip çalıştırmak.
- Modele soru + bağlam vererek RAG mantığında cevap ürettirmek.
- Verilen bağlamın, modelin ürettiği cevabı nasıl belirlediğini gözlemlemek.

## Kullanılan Araçlar

- **Python**
- **Hugging Face Transformers** (LLM'leri yükleyip çalıştırmak için kütüphane)
- **PyTorch** (modelin arka planda çalıştığı derin öğrenme kütüphanesi)
- **Google Colab** (çalışma ortamı)

## Yapılan Adımlar

1. Modelin `safetensors` ağırlıkları (weights: modelin öğrendiği milyonlarca parametre) indirildi ve belleğe yüklendi.
2. Modelin üretim ayarları (`generation_config.json`) yüklendi.
3. Modele bir **soru** ve bu soruyla ilişkili bir **bağlam** metni verildi.
4. Modelin, cevabı kendi ezberinden değil, verilen bağlamdan üretmesi sağlandı.

## Örnek Çalışma

| Alan | İçerik |
|------|--------|
| **Soru** | PyTorch'u kim geliştirdi? |
| **Verilen Bağlam** | Python, 1991 yılında Guido van Rossum tarafından geliştirilen, okunabilir bir programlama dilidir. |
| **Modelin Cevabı** | Guido van Rossum |

## Gözlem ve Çıkarılan Ders

Modelin verdiği cevap, PyTorch açısından **yanlıştır** (PyTorch aslında Meta / Facebook AI Research ekibi tarafından geliştirilmiştir; Guido van Rossum ise Python'un geliştiricisidir).

Ancak bu bir model hatası değildir: Model, **kendisine verilen bağlama sadık kalmıştır.** Bağlam Python hakkında olduğu için, model bu metinden çekebildiği tek ismi ("Guido van Rossum") cevap olarak vermiştir.

Bu deneme, RAG sistemlerinin temel kuralını net biçimde göstermektedir:

> **Model, kendisine getirilen bağlam kadar iyidir.** Yanlış veya alakasız bir bağlam getirilirse, doğru çalışan bir model bile yanlış cevap üretir ("garbage in, garbage out").

Yani asıl belirleyici kısım modelin kendisi değil, doğru bağlamı bulup getiren **retrieval** (soruya en uygun belgeyi arayıp seçme) adımıdır.

## Sonuç

RAG mantığında bir soru-cevap kurulumu başarıyla oluşturuldu ve çalıştırıldı. Deneme, bağlam kalitesinin cevabın doğruluğu üzerindeki belirleyici etkisini somut bir örnekle ortaya koydu. Bir sonraki adımda, soruya uygun (doğru) bir bağlam verilerek modelin cevabının nasıl değiştiği test edilebilir.
